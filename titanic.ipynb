```python
import pandas as pd
import numpy as np

col_name = ["PassengerId","Survived","Pclass","Name","Sex","Age","SibSp","Parch","Ticket","Fare","Cabin","Embarked"]
data = pd.read_csv('titanic.csv')

print(data.columns)
print("segglyuk")
print(data)
#dataset = pd.DataFrame(data["data"], columns = col_name)
data
```

```python
rows_to_drop = [] 

for key in data.columns:
    for i in data.index:
        val = data.at[i, key]
        if val is None or pd.isna(val):
            rows_to_drop.append(i)

rows_to_drop = list(set(rows_to_drop))


data = data.drop(rows_to_drop)
data
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
data['Sex'] = le.fit_transform(data['Sex'])
data
        
```

```python
data
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
columns_to_scale = ['Age','Fare']
scaled_col = ['Age_scale','Fare_scale']
scaled_data = data.copy()
scaled_data[scaled_col] = scaler.fit_transform(scaled_data[columns_to_scale])

data
```

```python
scaled_data
```

```python
import matplotlib.pyplot as plt
plt.hist(data['Pclass'], bins = 5, edgecolor = 'black')
plt.title('Pclass')
plt.xlabel('ertektartomany')
plt.ylabel('gyakorisag')
plt.show()

plt.hist(data['Sex'], bins = 5, edgecolor = 'black')
plt.xticks([0, 1], ['nő', 'férfi'])
plt.title('Nemek aranya')
plt.xlabel('ertektartomany')
plt.ylabel('gyakorisag')
plt.show()
```

```python
from scipy import stats
x = scaled_data['Age_scale']
y = scaled_data['Fare_scale']
slope, intercept, r, p, std_err = stats.linregress(x, y)

def myfunc(x):
  return slope * x + intercept

mymodel = list(map(myfunc, x))
plt.scatter(x,y)
plt.plot(x,mymodel)
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split


X = scaled_data[['Sex', 'Age', 'Pclass', 'Fare','SibSp']] 
Y = scaled_data['Survived']

X = X.fillna(X.mean())

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

model = LogisticRegression(max_iter=200)  
model.fit(X_train, y_train)

print("Pontosság:", model.score(X_test, y_test))
```
