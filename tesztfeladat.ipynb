```python
import os
import matplotlib
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_wine
import sklearn
from sklearn.preprocessing import MinMaxScaler
import numpy as np
import seaborn as sns
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import learning_curve
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score, ConfusionMatrixDisplay, classification_report
from sklearn.linear_model import LogisticRegression
```

```python
print(f"matplotlib version: {matplotlib.__version__}")
print(f"pandas version: {pd.__version__}")
print(f"sklearn version: {sklearn.__version__}")
print(f"numpy version: {np.__version__}")
print(f"seaborn version: {sns.__version__}")
col_name = ["Alcohol", "Malic acid", "Ash", "Alcalinity of ash", "Magnesium", "Total phenols", "Flavanoids", "Nonflavanoid phenols", "Proanthocyanins", "Color intensity", "Hue", "OD280/OD315 of diluted wines", "Proline"]
```

```python
#adat betöltése, oszlopnevek szerkesztése és Target hozzáadása
data = load_wine()
dataset = pd.DataFrame(data["data"], columns = col_name)
dataset.insert(13, "Target", data["target"])
col_name.append("Target")
```

```python
print(data['DESCR'])
```

```python
#azon sorok kitörlése, amelynek valamelyik cellája üres, nem számot vagy outliert tartalmaz. Nem akartam szennyezni az adathalmazt, így töröltem
for key in col_name:
    if key == "Target":
        for i in dataset[key].keys():
            
            if dataset[key][i]==None or dataset[key][i]==np.nan:
                dataset = dataset.drop(i)
        continue
    atlag=dataset[key].mean()
#6-szigma szabályt alkalmaztam 
    also_hatar=atlag-6*dataset[key].std()
    felso_hatar=atlag+6*dataset[key].std()
    
    for i in dataset[key].keys():
        if also_hatar> dataset[key][i] or felso_hatar<dataset[key][i] or dataset[key][i]==None or dataset[key][i]==np.nan:
            dataset = dataset.drop(i)
```

```python
#duplikált sorok törlése
dataset = dataset.drop_duplicates()
dataset
```

```python
#Target oszlop kitörlése a későbbi normalizálás miatt. Azt nem normalizálom
target = list(dataset["Target"])
dataset = dataset.drop("Target", axis='columns')
#statisztikai információk
dataset.describe()
```

```python
#adattranszformáció, normalizálás
scaler = MinMaxScaler()
scaler.fit(dataset)

scaled_data = scaler.transform(dataset)
```

```python
#adathalmaz öszeállítása és Target visszahelyezése
dataset_scaled = pd.DataFrame(scaled_data , columns = col_name[:-1])
dataset_scaled.insert(13, "Target", target)
dataset.insert(13, "Target", target)
```

```python
#klaszterezés
j_kezdo=1
i_kezdo=0
for i in range(i_kezdo, len(col_name)-1):
    for j in range(j_kezdo, len(col_name)-1):
            X = np.array(dataset_scaled[[col_name[i], col_name[j]]])
            kmeans = KMeans(n_clusters=3)
            kmeans.fit(X)
            centroids = kmeans.cluster_centers_
            labels = kmeans.labels_
            plt.scatter(X[:, 0], X[:, 1], c=labels, s=50, cmap='viridis')
            plt.scatter(centroids[:, 0], centroids[:, 1], s=200, c='red', marker='x')
            plt.title(f"K-means clustering: {col_name[i]} vs {col_name[j]}")
            plt.xlabel(col_name[i])
            plt.ylabel(col_name[j])
            plt.show()
    j_kezdo+=1
```

```python
y = list(dataset_scaled["Target"])
X = dataset_scaled.drop("Target", axis='columns')
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=777)

# Lineáris regressziós modell létrehozása és illesztése
model = LogisticRegression(max_iter=10000, multi_class='ovr')
```

```python
stratified_kf = StratifiedKFold(n_splits=5)

train_sizes, train_scores, test_scores = learning_curve(
    model, pd.concat([X_train, X_test], ignore_index=True), y_train+y_test, cv=stratified_kf, scoring='accuracy', train_sizes=np.linspace(0.1, 1.0, 10)
)
```

```python
train_mean = -np.mean(train_scores, axis=1)
train_std = np.std(train_scores, axis=1)
test_mean = -np.mean(test_scores, axis=1)
test_std = np.std(test_scores, axis=1)

# Tanulási görbék vizualizálása
plt.figure(figsize=(10, 6))
plt.plot(train_sizes, train_mean, 'o-', color='blue', label='Train score (MSE)')

# Hibacsíkok hozzáadása
plt.fill_between(train_sizes, train_mean - train_std, train_mean + train_std, alpha=0.1, color='blue')

plt.xlabel('Training Size')
plt.ylabel('Mean Squared Error')
plt.title('Tanulási Görbék: Költségfüggvény')
plt.legend(loc='best')
plt.grid(True)
plt.show()
```

```python
model.fit(X_train, y_train)
```

```python
 model.predict(X_test)
```

```python
y_pred = model.predict(X_test)# Valószínűségek a pozitív osztályra

# Konfúziós mátrix számítása
conf_matrix = confusion_matrix(y_test, y_pred)
print(f'Confusion Matrix:\n{conf_matrix}')

# Konfúziós mátrix vizualizálása
disp = ConfusionMatrixDisplay(conf_matrix, display_labels=data.target_names)
disp.plot(cmap=plt.cm.Blues)
plt.title('Confusion Matrix')
plt.show()
```

```python
print(classification_report(y_test, y_pred))
```
